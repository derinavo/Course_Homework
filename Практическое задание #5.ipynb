Считайте датасет Housing.csv с помощью pandas

```python
housing = pd.read_csv('Housing.csv')
housing.info(memory_usage='deep')
housing.head()
```

Для каждого столбца определите тип данных с меньшим потреблением памяти - 2 балла

```python
print(housing.agg(['min', 'max']).loc[:, ['price', 'area', 'bedrooms', 'bathrooms', 'stories', 'parking']])
print(housing.nunique())
```

Поменяйте типы столбцов и сравните память до и после - 1 балл

```python
memory_before = housing.memory_usage(deep=True).sum()

housing_optimized = housing.copy()
for column, dtype in {'price': np.int32, 'area': np.int16,
                      'bedrooms': np.int8, 'bathrooms': np.int8,
                      'stories': np.int8, 'parking': np.int8}.items():
    housing_optimized[column] = housing_optimized[column].astype(dtype)

for column in ['mainroad', 'guestroom', 'basement', 'hotwaterheating',
               'airconditioning', 'prefarea', 'furnishingstatus']:
    housing_optimized[column] = housing_optimized[column].astype('category')

memory_after = housing_optimized.memory_usage(deep=True).sum()
print(housing_optimized.dtypes)
print(f'До: {memory_before:,} байт')
print(f'После: {memory_after:,} байт')
print(f'Экономия: {(1 - memory_after / memory_before) * 100:.1f}%')
```
