```python
import requests
import pandas as pd
import psycopg2
```

Получить информацию обо всех книгах и сохранить в pd.DataFrame - 1 балл

```python
url = 'https://www.anapioficeandfire.com/api'
books = []
page = 1
while True:
    items = requests.get(
        f'{url}/books?page={page}&pageSize=50'
    ).json()
    if not items:
        break
    books += items
    page += 1

df_books = pd.DataFrame(books)
print('Книг:', len(df_books))
df_books.head()
```

Получить информацию обо всех домах Вестероса (Houses) и сохранить в pd.DataFrame - 1 балл

```python
houses = []
page = 1
while True:
    items = requests.get(
        f'{url}/houses?page={page}&pageSize=50'
    ).json()
    if not items:
        break
    houses += items
    page += 1

df_houses = pd.DataFrame(houses)
print('Домов:', len(df_houses))
df_houses.head()
```

Получить информацию обо всех домах Вестероса, у которых есть девиз, и сохранить в pd.DataFrame - 2 балла

```python
houses_words = []
page = 1
while True:
    items = requests.get(
        f'{url}/houses?hasWords=true&page={page}&pageSize=50'
    ).json()
    if not items:
        break
    houses_words += items
    page += 1

df_house_words = pd.DataFrame(houses_words)
print('Домов с девизом:', len(df_house_words))
df_house_words[['name', 'words']].head()
```
