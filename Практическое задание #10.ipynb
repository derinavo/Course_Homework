```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import joblib
import pickle
import streamlit as st
```

Обучить модель прогнозирования стоимости недвижимости - 1

```python
data = pd.read_csv('realty_data.csv')
data.head()
```

```python
target_column = 'price'
feature_columns = ['total_square', 'floor']

train_data = data[feature_columns + [target_column]].dropna()
X = train_data[feature_columns]
y = train_data[target_column]

model = LinearRegression()
model.fit(X, y)
model_path = 'priceM.pkl'
with open(model_path, 'wb') as file:
    pickle.dump(model, file)
```

Код для получения предсказания обученной моделью - 1 балл

```python
inputDF = X.head(1)
prediction = model.predict(inputDF)[0]
print(f'Прогноз: {prediction:,.2f}')
```
